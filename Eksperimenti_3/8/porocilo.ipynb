Vaja 8 - Stoječe valovanje

Vrvico smo napeli z silo/napetostjo F. In jo nihali z frekvenco $\nu = 100 hz$.
Iskali smo $n$ stoječa valovanja pri neki napetosti $F$:

$$
\begin{array}{|c|c|}
\hline
n & F [N] \\
\hline
1 & 6.8 \pm 0.1 \\
2 & 2.5 \pm 0.1 \\
3 & 1.41 \pm 0.02 \\
4 & 0.9 \pm 0.02 \\
5 & 0.55 \pm 0.02 \\
\hline
\end{array}
$$

Izmerili smo tudi dolžinsko gostoto:
$$
\begin{array}{}
    m = (0.49 \pm 0.01)g \\
    l = (2.973 \pm 0.003)m \\
\end{array}
$$
Ker je hitrost valovanja $c = \sqrt{F\frac{l}{m}}$ Lahko narišemo graf $c^2(F)$:



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd

plt.style.use('bmh')

df = pd.read_csv("data.csv", delimiter=";", decimal=".")

x = df.iloc[:,1].to_numpy()      # F
y = df.iloc[:,3].to_numpy()      # C^2 daj 4 za C
x_err = df.iloc[:,5].to_numpy()
y_err = df.iloc[:,8].to_numpy()  # 8 daj 9 za c err

def f(x, a, b):
    return a*x + b

params, cov = curve_fit(f, x, y)
a, b = params
errors = np.sqrt(np.diag(cov))

plt.errorbar(
    x, y,
    yerr=y_err,
    xerr=x_err,
    fmt='s',  # square markers
    markersize=4,  # smaller dots
    color='#1f3b73',  # dark blue (markers + line)
    ecolor='#444444',  # dark grey error bars
    elinewidth=1,  # thin error lines
    capsize=3,  # "T" caps size
    capthick=1,  # thickness of caps
    markeredgewidth=1,
    linestyle='none',  # no connecting lines
    label='meritve'
)


x_fit = np.linspace(min(x), max(x), 100)
plt.plot(x_fit, f(x_fit, a, b), label=f'fit: y={a:.3e}x+{b:.3e}', linewidth=0.8)

plt.legend()
plt.xlabel(r"$F\,[N]$")
plt.ylabel(r"$C^2\,[m^s/s^2]$")
plt.savefig("graf1test.png", dpi=300)
plt.show()

print(f"k = {a:.4e} ± {errors[0]:.4e}")
print(f"n = {b:.4e} ± {errors[1]:.4e}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd

plt.style.use('bmh')

df = pd.read_csv("data_vzmet.csv", delimiter=";", decimal=".")

x = df.iloc[:,0].to_numpy()      # F
y = df.iloc[:,1].to_numpy()      # C^2 daj 4 za C

def f(x, a, b):
    return a*x + b

params, cov = curve_fit(f, x, y)
a, b = params
errors = np.sqrt(np.diag(cov))

plt.errorbar(
    x, y,
    yerr=0,
    xerr=0,
    fmt='s',  # square markers
    markersize=4,  # smaller dots
    color='#1f3b73',  # dark blue (markers + line)
    ecolor='#444444',  # dark grey error bars
    elinewidth=1,  # thin error lines
    capsize=3,  # "T" caps size
    capthick=1,  # thickness of caps
    markeredgewidth=1,
    linestyle='none',  # no connecting lines
    label='meritve'
)


x_fit = np.linspace(min(x), max(x), 100)
plt.plot(x_fit, f(x_fit, a, b), label=f'fit: y={a:.3e}x+{b:.3e}', linewidth=0.8)

plt.legend()
plt.xlabel(r"$x\,[m]$")
plt.ylabel(r"$F\,[N]$")
plt.savefig("graf1test.png", dpi=300)
plt.show()

print(f"k = {a:.4e} ± {errors[0]:.4e}")
print(f"n = {b:.4e} ± {errors[1]:.4e}")
```
