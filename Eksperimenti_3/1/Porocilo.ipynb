3. Določanje fokusa z krivinskim polmerom

Meritve:
$$
v_{kriv} = (0.68 \pm 0.01) mm
$$

Polmer vsake "nogice" sferometra:
$$
\begin{array}{}
    \rho_{1} = (1.265  \pm 0.005) cm \\
    \rho_{2} = (1.320  \pm 0.005) cm \\
    \rho_{3} = (1.375  \pm 0.005) cm \\
\end{array}
$$

$$
\bar{\rho} = (1.32 \pm 00.1) cm \\
$$
Lomni količnik:
$$
n_{stekla} = 1.4 \pm 0.01
$$

Izračun polera:
$$
r = \frac{\rho^2 - v^2}{2v} = 12.8 (1 \pm 0.05)cm = (12.8 \pm 0.6)cm
$$



Izračun fokusa:

$$
f = ((n-1)(\frac{1}{r_1} - \frac{1}{r_2}))^{-1} = 16.0(1 \pm 0.1)cm = (16\pm 2)cm
$$

4. Določitev fokusa sistema leč

Meritve:
$$
\begin{array}{|c|c|}
\hline
l [cm] & s [cm] \\
\hline
79.0 \pm 0.4 & 15.0 \pm 0.4 \\
79.0 \pm 0.4 & 16.2 \pm 0.4 \\
\hline
\end{array}
$$

$$
d = (18.0 \pm 0.4) cm
$$

Izračun skupne goriščne razdalje:

$$
f_1=\frac{l^2 - s^2}{4l} = 19.0 (1 \pm 0.09)cm = (19 \pm 2) cm
$$

$$
f_2=\frac{l^2 - s^2}{4l} = 18.9 (1 \pm0.09)cm = (19 \pm 2) cm
$$

Ter goriščna razdalja konkavne leče:

$$
\frac{1}{f} = \frac{1}{f_1} + \frac{1}{f_2} - \frac{d}{f_1 f_2}
$$
Izpostavimo f_2, dobimo:

$$
f_2 = (1 - \frac{d}{f_1}) (\frac{1}{f} - \frac{1}{f_1})^-1 = 12.7 (1 \pm 0.1)cm = (12 \pm 1)cm
$$
