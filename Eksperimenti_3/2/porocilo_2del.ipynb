Spekter plinov:

Opazovali smo spektre plinov skozi 300 rež/mm uklonsko mrežico. Na razdalji $ (L = 20.4 \pm 0.2)cm$ od svetilke.

Zaznali smo sledeča vrha: moder pri $x_m = 3.0 cm \pm 0.1$ in rdeč $x_r = 4.1 cm \pm 0.1$.

Valovna dolžina teh svetlob sta bile:


$$
\\
\begin{array}{}
    a*sin(\beta) = \lambda \\
    a = 3.33 \mu m \\
    \beta_m = tan^{-1}(x_m/l) = 8.4^\circ \\
    \beta_r = tan^{-1}(x_r/l) = 11.4^\circ \\
    \lambda_m \approx 486 nm\\
    \lambda_r \approx 650 nm\\
\end{array}{}
$$
